```python
def clean_to_float(series):
    s = series.astype(str).str.strip()
    
    s = s.str.replace(r'[\$\€\£%]', '', regex=True)
    s = s.str.replace(',', '', regex=True)
    s = s.str.replace(r'\s+', '', regex=True)
    
    s = s.replace({'': None, 'nan': None, 'None': None})
    return pd.to_numeric(s, errors='coerce')
```

```python
money_cols = ['per_capita_income', 'yearly_income', 'total_debt']
for c in money_cols:
    if c in users.columns:
        newc =  "num_" + c
        users[newc] = clean_to_float(users[c])   # كل عمليات التنظيف تحصل هنا
        print(f"{c} -> {newc}: NaN count = {users[newc].isna().sum()} / {len(users)}")
    else:
        print(f"Skipped {c} (not present)")
```

```python
num_cols = ['num_per_capita_income', 'num_yearly_income', 'num_total_debt', 'credit_score', 'num_credit_cards']
present = [c for c in num_cols if c in users.columns]
print("Numeric columns present:", present)
display(users[present].describe(percentiles=[0.05,0.25,0.5,0.75,0.95,0.99]))
```

```python
if 'num_total_debt' in users.columns and 'num_yearly_income' in users.columns:
    users['debt_to_income'] = users['num_total_debt'] / users['num_yearly_income'].replace({0: np.nan})
    users['debt_to_income'] = users['debt_to_income'].replace([np.inf, -np.inf], np.nan).fillna(0)
    print("Created debt_to_income; sample stats:")
    display(users['debt_to_income'].describe(percentiles=[0.05,0.25,0.5,0.75,0.95,0.99]))
else:
    print("Required columns missing for debt_to_income.")
```

```python
# credit_score: نحوله لرقم ونعمل clipping لو برا النطاق الشائع
if 'credit_score' in users.columns:
    users['credit_score'] = pd.to_numeric(users['credit_score'], errors='coerce')
    users['credit_score'] = users['credit_score'].clip(lower=300, upper=850)
    print( "credit_score cleaned")
```

```python
# num_credit_cards: نحوله لرقم، نخلى الفاضي بـ0، ونخليه int موجب
if 'num_credit_cards' in users.columns:
    users['num_credit_cards'] = pd.to_numeric(users['num_credit_cards'], errors='coerce').fillna(0)
    users['num_credit_cards'] = users['num_credit_cards'].round().astype(int).clip(lower=0)
```

```python
df_users = pd.read_sql("SELECT * FROM users_data_clean", conn)
df_cards = pd.read_sql("SELECT * FROM Cards_data_clean", conn)
df_transactions = pd.read_sql("SELECT * FROM Transactions_data", conn)


print("Users Table:")
print(df_users.head(), "\n")

print("Cards Table:")
print(df_cards.head(), "\n")

print("Transactions Table:")
print(df_transactions.head(), "\n")

```

```python

queries = {
    "df_num_cards": """
        SELECT u.id AS client_id, COUNT(c.id) AS num_cards
        FROM users_data_clean u
        LEFT JOIN cards_data_clean c ON u.id = c.client_id
        GROUP BY u.id
        ORDER BY num_cards DESC""",
    
    "df_card_types": """
        SELECT u.id AS client_id, c.card_type, COUNT(*) AS count_per_type
        FROM users_data_clean u
        JOIN cards_data_clean c ON u.id = c.client_id
        GROUP BY u.id, c.card_type
        ORDER BY u.id, count_per_type DESC""",
    
    "df_avg_amount": """
        SELECT c.id AS card_id, c.card_type, AVG(t.amount) AS avg_amount
        FROM cards_data_clean c
        LEFT JOIN transactions_data t ON c.id = t.card_id
        GROUP BY c.id, c.card_type
        ORDER BY avg_amount DESC""",
    
    "df_total_analysis": """
        SELECT u.id AS client_id, u.gender, u.birth_year,
               COUNT(DISTINCT c.id) AS num_cards,
               COUNT(t.id) AS total_transactions,
               SUM(CASE WHEN t.is_fraud = 1 THEN 1 ELSE 0 END) AS fraud_count
        FROM users_data_clean u
        LEFT JOIN cards_data_clean c ON u.id = c.client_id
        LEFT JOIN transactions_data t ON c.id = t.card_id
        GROUP BY u.id, u.gender, u.birth_year
        ORDER BY fraud_count DESC""",
    
    "df_chip_vs_swipe": """
        SELECT c.card_type, t.use_chip, COUNT(*) AS num_transactions
        FROM cards_data_clean c
        JOIN transactions_data t ON c.id = t.card_id
        GROUP BY c.card_type, t.use_chip
        ORDER BY c.card_type, num_transactions DESC""",
    
    "df_fraud_by_card_type": """
        SELECT c.card_type, t.is_fraud, COUNT(*) AS count_transactions
        FROM cards_data_clean c
        JOIN transactions_data t ON c.id = t.card_id
        GROUP BY c.card_type, t.is_fraud
        ORDER BY c.card_type, t.is_fraud DESC""",
    
    "df_total_amount": """
        SELECT u.id AS client_id, c.card_type, SUM(t.amount) AS total_amount
        FROM users_data_clean u
        JOIN cards_data_clean c ON u.id = c.client_id
        JOIN transactions_data t ON c.id = t.card_id
        GROUP BY u.id, c.card_type
        ORDER BY total_amount DESC""",
    
    "df_top_users_fraud": """
        SELECT u.id AS client_id, c.id AS card_id, c.card_type,
               COUNT(t.id) AS total_transactions,
               SUM(CASE WHEN t.is_fraud = 1 THEN 1 ELSE 0 END) AS fraud_count
        FROM users_data_clean u
        JOIN cards_data_clean c ON u.id = c.client_id
        JOIN transactions_data t ON c.id = t.card_id
        GROUP BY u.id,  c.id, c.card_type
        ORDER BY fraud_count DESC"""
}


dfs = {}
for name, query in queries.items():
    dfs[name] = pd.read_sql(query, conn)
    print(f"{name} loaded: {dfs[name].shape} rows")
```

```python
# VISUALIZATION
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
sns.set(style="whitegrid") 
palette = sns.color_palette("Set2") 
```

```python
top_users = dfs['df_num_cards'].nlargest(20, 'num_cards')

plt.figure(figsize=(12,6))
sns.barplot(data=top_users, x='client_id', y='num_cards', palette="Set2")
plt.xticks(rotation=45)
plt.title("Top 20 Users by Number of Cards", fontsize=16, fontweight='bold')
plt.xlabel("Client ID", fontsize=12)
plt.ylabel("Number of Cards", fontsize=12)
plt.show()
```
